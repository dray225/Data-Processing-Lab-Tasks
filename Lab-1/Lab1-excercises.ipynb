### 01 • Simple Calculator

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

summ = num1 + num2
diff = num1 - num2
prod = num1 * num2

#show results

print("--- Results ---")
print(f"Sum: {summ}")
print(f"Difference: {diff}")
print(f"Product: {prod}")

if num2 != 0:
    quotient = num1 / num2
    print(f"Quotient: {quotient}")
else:
    print("Quotient: Can't divide by 0")
```

### 02 • Simple Salary Calculator

```python
name = input("Enter employee name: ")
base_salary = float(input("Enter base salary: "))
allowance = float(input("Enter allowance: "))

gross_salary = base_salary + allowance

#determine taxrate
if base_salary <= 30000:
    tax_rate = 0.0
elif base_salary <= 50000:
    tax_rate = 0.05
elif base_salary <= 80000:
    tax_rate = 0.10
else:
    tax_rate = 0.15

#calc tax amount
tax_amount = gross_salary * tax_rate
net_salary = gross_salary - tax_amount

#show summary
print(f"---- Payslip ----")
print(f"Name: {name}")
print(f"Gross Salary: {gross_salary}")
print(f"Tax Rate: {tax_rate * 100}%")
print(f"Tax Amount: {tax_amount}")
print(f"Net Salary: {net_salary}")
```

### 03 • Student Performance Analyzer

```python
name = input("Enter student name: ")

marks = []
passed_count = 0

for i in range(5):
    mark = float(input(f"Enter mark for course {i+1}: ")) #since i starts from 0
    marks.append(mark)

    if mark >= 50:
        passed_count = passed_count + 1


total = sum(marks)
average = total/5
highest = max(marks)
lowest = min(marks)

#determine class
if average >= 80:
    perf = "Excellent"
elif average >= 70:
    perf = "Good"
elif average >= 60:
    perf = "Satisfactory"
elif average >= 50:
    perf = "Pass"
else:
    perf = "Needs Improvement"

#show results
print("--- Student Performance Report ---")
print(f"Name: {name}")
print(f"Total Marks: {total}")
print(f"Average Marks: {average}")
print(f"Highest Mark: {highest}")
print(f"Lowest Mark: {lowest}")
print(f"Passed Courses: {passed_count} / 5")
print(f"Performance: {perf}")
```

### 04 • Train Ticket Calculator

```python
#assuming location names and fare prices

print("Available Destinations: A ($100), B ($200), C ($300)")
destination = input("Enter destination: ").strip().upper()

if destination == "A":
    base_fare = 100
elif destination == "B":
    base_fare = 200
elif destination == "C":
    base_fare = 300
else:
    print("Invalid destination")
    exit()

#input passenger details
passenger_name = input("Enter passenger name: ")
ticket_count = int(input("Enter number of tickets: "))
passenger_type = input("Enter passenger type (regular/child/student/senior): ").strip().upper()

#determine discount
if passenger_type == "CHILD":
    discount = 0.50
elif passenger_type == "STUDENT":
    discount = 0.30
elif passenger_type == "SENIOR":
    discount = 0.40
else:
    discount = 0.00 #regular

#calc fare
fare_per_ticket = base_fare - (base_fare * discount)
total_fare = fare_per_ticket * ticket_count

#show results
print("\n-------TRAIN TICKET-------")
print(f"Passenger Name : {passenger_name}")
print(f"Destination    : {destination}")
print(f"Passenger Type : {passenger_type}")
print(f"Ticket Count   : {ticket_count}")
print(f"Fare per Ticket: ${fare_per_ticket}")
print("------------------------------")
print(f"TOTAL FARE     : ${total_fare}")
```
