Import libraries

```python
import pandas as pd
import requests
```

Send api request

```python
url = (
"https://api.open-meteo.com/v1/forecast"
"?latitude=23.8103&longitude=90.4125"
"&daily=temperature_2m_max,temperature_2m_min,precipitation_sum"
"&timezone=Asia%2FDhaka&forecast_days=7"
)

response = requests.get(url, timeout=20)
print("Status code:", response.status_code)
response.raise_for_status()

#success -> status code 200
```

Exctract json

```python
data = response.json()

daily_data = data["daily"]
```

Create df

```python
df_weather = pd.DataFrame(daily_data)

# Rename columns for clarity
df_weather = df_weather.rename(columns={
    "time": "date",
    "temperature_2m_max": "max_temperature",
    "temperature_2m_min": "min_temperature",
    "precipitation_sum": "precipitation"
})

display(df_weather.head())
```

Inspect dataset

```python
print(f"Dimensions: {df_weather.shape}")
print(f"Columns: {df_weather.columns.tolist()}")
print(f"\nData types:\n\n{df_weather.dtypes}")
print(f"\nMissing values:\n\n{df_weather.isnull().sum()}")
```
