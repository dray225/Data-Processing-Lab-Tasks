Import libraries

```python
import pandas as pd
import requests
```

Send api request

```python
url = (
"https://api.worldbank.org/v2/country/BGD/"
"indicator/SP.POP.TOTL"
"?format=json&per_page=100")

response = requests.get(url, timeout=20)
print("Status code:", response.status_code)
response.raise_for_status()

#success -> status code 200
```

Exctract json

```python
data = response.json()

metadata = data[0]
records = data[1]

print(f"total records: {metadata['total']}")
print(f"records received: {len(records)}")
print(records[0])
```

Create df

```python
df_population = pd.DataFrame(records)

df_population = df_population[["date", "value"]].rename(columns={"date": "year","value": "population"})

display(df_population.head())
```

Inspect dataset

```python
print(f"# Dimensions: {df_population.shape}")
print(f"# Columns: {df_population.columns.tolist()}")
print(f"\n# Data types:\n\n{df_population.dtypes}")
print(f"\n# Missing values:\n\n{df_population.isnull().sum()}")
```

Save Dataset

```python
df_population.to_csv("bangladesh_population.csv", index=False) #index true adds an index column
```
