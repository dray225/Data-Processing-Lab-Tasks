### 02 • Course Marks with a List

```python
name = input("Enter student name: ")

marks = []
passed_count = 0

for i in range(5):
    mark = float(input(f"Enter mark for course {i+1}: ")) #since i starts from 0
    marks.append(mark)

    if mark >= 50:
        passed_count = passed_count + 1


total = sum(marks)
average = total/5
highest = max(marks)
lowest = min(marks)

#determine class
if average >= 80:
    perf = "Excellent"
elif average >= 70:
    perf = "Good"
elif average >= 60:
    perf = "Satisfactory"
elif average >= 50:
    perf = "Pass"
else:
    perf = "Needs Improvement"

#show results
print("--- Student Performance Report ---")
print(f"Name: {name}")
print(f"Total Marks: {total}")
print(f"Average Marks: {average}")
print(f"Highest Mark: {highest}")
print(f"Lowest Mark: {lowest}")
print(f"Passed Courses: {passed_count} / 5")
print(f"Performance: {perf}")
```

### 06 • Grade Calculator Function

```python
def calculate_grade(marks):

    average = sum(marks) / len(marks)
    
    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"

mark_arr = [85, 72, 90, 68, 78]
final_grade = calculate_grade(mark_arr) #returns string grade

#show results
avg_score = sum(mark_arr) / len(mark_arr)
print("----Grade Info----")
print(f"Marks: {mark_arr}")
print(f"Average: {avg_score}")
print(f"Grade: {final_grade}")
```

### 07 • Shopping Cart Processor

```python
#sample data for 7,8
students = [
    {"name": "Rahim", "department": "CSE", "marks": 85, "attendance": 92},
    {"name": "Karim", "department": "CSE", "marks": 48, "attendance": 76},
    {"name": "Sara", "department": "DS", "marks": 91, "attendance": 95},
    {"name": "Nadia", "department": "DS", "marks": 67, "attendance": 84},
    {"name": "Hasan", "department": "CSE", "marks": 73, "attendance": 78}
]
products = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1}
]
```

```python
total_cost = 0
for i in products:
    category_cost = i["price"] * i["quantity"]
    total_cost += category_cost

def apply_discount(total):
    if total >= 2000:
        return total * 0.90
    else:
        return total

final_price = apply_discount(total_cost)

print("---Shopping Cart Info---")
print(f"Total Cost: ${total_cost}")
print(f"Discounted Price: ${final_price}")
```

### 08 • Mini Data Processor 

```python
def calculate_average_marks(student_arr):
    total_marks = 0
    for student_dict in student_arr:
        total_marks += student_dict["marks"]
    return total_marks / len(student_arr)

def find_top_student(student_arr):
    top_dict = student_arr[0]
    for student_dict in student_arr:
        if student_dict["marks"] > top_dict["marks"]:
            top_dict = student_dict
    return top_dict

def count_low_attendance(student_arr):
    limit = 80
    count = 0
    for student_dict in student_arr:
        if student_dict["attendance"] < limit:
            count += 1
    return count

avg_marks = calculate_average_marks(students)
top_dict = find_top_student(students)
low_attendance_count = count_low_attendance(students)

#show results
print("--- STUDENT PERFORMANCE ---")
print(f"Average Marks: {avg_marks}")
print(f"Top Performer: {top_dict['name']} ({top_dict['department']}) with {top_dict['marks']} marks")
print(f"Students with Attendance < 80%: {low_attendance_count}")
```
